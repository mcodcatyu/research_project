### Here we generate small simple dummy train and test datasets for users to try out the  GUI 

**We strip the real records from the original file to keep only its template, then paste our synthetic data into it.**

*So if you'd like to test with a larger dataset, you can you can generate more data here and copy-paste it into the provided train/test files*

```python
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

def generate_data (num_rows = 50, start_time = "1994-02-16 19:20:00"):
    """
        Generate dummy data for users to try out the GUI
        Args: 
            num_rows (int, optional): produced data row, default is 50 rows
            start_time (str, optional): data start time, format is "YYYY-MM-DD HH:MM:SS". Default is 1994-02-16 19:20:00
        Returns:
            str: A multi-column formatted text string concatenated with newline characters (\n)
    """
    current_dt = datetime.strptime(start_time, "%Y-%m-%d %H:%M:%S")
    rows = []

    for i in range(num_rows):
        date_str = current_dt.strftime("%y%m%d")
        time_str = current_dt.strftime("%H%M%S")

        is_std = np.random.rand() < 0.1
        sample_type = "std" if is_std else "air"
        height = "10m" 

        # Generate data 
        v1 = f"{np.random.uniform(22.0, 35.0): 6.2f}" # tamb
        v2 = f"{np.random.uniform(825.0, 848.0):6.1f}" # pflow
        v3 = f"{np.random.uniform(748.0, 761.0):6.2f}" # psamp
        v4 = f"{np.random.uniform(748.0, 761.0): 6.1f}" # pamb

        is_zero_state = (not is_std) and (np.random.rand() < 0.25)
        """
        c1 : CH4_ht, c2: CH4_area, c3: CH4_skew, f1: CH4_start_time, f2:CH4_end_time, f3:CH4_start_level, f4:CH4_end_level, f5:CH4_C
        """
        if is_zero_state:
            c1 = f"{0:8d}"
            c2 = f"{0:9d}"
            c3 = f"{0.0:5.2f}"
            f1 = f"{0.0:10.2f}"
            f2 = f"{0.0:8.2f}"
            f3 = f"{0.0:11.2f}"
            f4 = f"{0.0:10.2f}"
            f5 = f"{0.0:7.2f}"
            

        else:
            c1 = f"{np.random.randint(1000, 18000):8d}"
            c2 = f"{np.random.randint(70000, 1800000):9d}"
            c3 = f"{np.random.uniform(0.01, 0.35):5.2f}"
            f1 = f"{np.random.uniform(0.0, 2.8): 10.2f}"
            f2 = f"{np.random.uniform(0.5, 17.5):8.2f}"
            f3 = f"{np.random.uniform(0.0, 320.0): 11.2f}"
            f4 = f"{np.random.uniform(0.0, 1200.0): 10.2f}"
            f5 = f"{np.random.uniform(1600.0, 1800.0):5.2f}"

        # Generate flag 
        flag1 = ("F" if (is_zero_state or is_std or np.random.rand() < 0.02) else " ")
        flag2 = "F" if (np.random.rand() < 0.005) else " "
        row = (
            f"{date_str} {time_str}           {sample_type:>3s}         {height:>4s}        G-024    3    10 40.00{v1}   nan {v2} {v3}   nan {v4}    0.0  0.00 {c1} {c2} {c3} {f1}  {f2}{f3}{f4}      nan      nan      nan     nan     nan     nan      nan      nan      nan     nan     nan     {f5}       nan      nan 1692.489       {flag1}      {flag2}             ".rstrip()   
        )
        rows.append(row)
        current_dt += timedelta(minutes=20)
    return "\n".join(rows)

print(generate_data(500)) # generate train data
```
