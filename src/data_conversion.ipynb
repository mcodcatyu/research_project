### Preprocessing

```python
real_data_mhd ["CH4_flag"] = real_data_mhd ["CH4_flag"].fillna(" ")
real_data_mhd ["CH4_flag_ht"] = real_data_mhd ["CH4_flag_ht"].fillna(" ")
real_data_mhd ["CH4_flag_a"] = real_data_mhd ["CH4_flag_a"].fillna(" ")
```

```python
mapping= {" ":0, 
          "x":0, 
          "*":1,
          "F":1, 
          "B":0, 
        }

real_data_mhd ["CH4_flag_ht_bin"] = real_data_mhd ["CH4_flag_ht"].map(mapping)
real_data_mhd ["CH4_flag_a_bin"] = real_data_mhd ["CH4_flag_a"].map(mapping)
```

```python
real_data_mhd.head()
```

```python
real_data_mhd['label1'] = (
    (real_data_mhd['CH4_flag_ht_bin'] == 1) |
    (real_data_mhd['CH4_flag_a_bin'] == 1)
).astype(int)
```

```python
real_data_mhd['date'] = real_data_mhd['date'].astype(str).str.zfill(6)
real_data_mhd['time'] = real_data_mhd['time'].astype(str).str.zfill(6)

real_data_mhd['datetime_str'] = real_data_mhd['date'] + real_data_mhd['time']

real_data_mhd['datetime'] = pd.to_datetime(real_data_mhd['datetime_str'], format='%y%m%d%H%M%S')

real_data_mhd.head(20)
real_data_mhd= real_data_mhd.drop(columns=['datetime_str','CH4_flag_a_bin', 'CH4_flag_a_bin'])
```

```python
real_data_mhd['datetime'] = pd.to_datetime(real_data_mhd['datetime'])

real_data_mhd['year'] =real_data_mhd['datetime'].dt.year
real_data_mhd['month']=real_data_mhd['datetime'].dt.month
real_data_mhd['day']=real_data_mhd['datetime'].dt.day
real_data_mhd['hour']=real_data_mhd['datetime'].dt.hour

real_data_mhd['minute']=real_data_mhd['datetime'].dt.minute
real_data_mhd['seconds']=real_data_mhd['datetime'].dt.second
```

```python
real_data_mhd.set_index('datetime')
real_data_mhd.head()
```

```python
real_data_mhd = real_data_mhd[((real_data_mhd['type'] == 'air')|(real_data_mhd['type']=='std'))]
real_data_mhd ['is_air'] = np.where(real_data_mhd ['type']=='air', 1, 0)
real_data_mhd ['is_std'] = np.where(real_data_mhd ['type']=='std', 1, 0)
```

```python
real_data_mhd.head()
```

```python
real_data_mhd = real_data_mhd.set_index('datetime').sort_index()
```
