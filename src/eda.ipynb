```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.max_rows', 1000)
pd.set_option('display.max_columns', 1000);
df = pd.read_csv('../data/processed/mhd_ch4_cnan_v1.csv', index_col= 'datetime')
df = df.drop(df[df['year']==2026].index)
```

### Yearly Data Count and Label Composition (Label 0 (Normal) vs. Label 1 (Anomaly))

```python
df_counts = pd.crosstab(df['year'], df['label1'])

ax = df_counts.plot(kind='bar', stacked=True, figsize=(20, 12))

plt.title('Count of Labels(label1) by Year')
plt.xlabel('Year')
plt.ylabel('Count')

plt.xticks(rotation=0)
plt.legend(title='Label')
plt.show()
```

```python
anomaly = (df['label1'] ==1).sum()
normal = (df['label1'] == 0).sum()
total = len(df['label1'])

print(f'Number of normal data points: {normal}')
print(f'Number of anomalous data points: {anomaly}')
print(f'Anomaly rate: {(anomaly/total)*100:.2f}%')
```

## Trend of Methane (CH4) Mole Fraction

```python

plt.figure(figsize=(10,6))

plt.plot(
    df['year'],
    df['CH4_C'],
    color = 'royalblue',
    marker = '+',
    alpha = 0.3
)

plt.xlabel('Year'),
plt.ylabel('Mole fraction (1e-9)')
plt.title('Mole fraction of CH4 Distribution by Year')
plt.grid()
```

### Relationship between the original anomaly flag and the characteristic distribution (1000-point random sampling) -`flag_a` and `flag_ht` 

** Flag definition **

* `*` **Auto-rejected** : automatically rejected according to predefined thresholds specified within GCWerks

* `F` **manually rejected** :reason may be given in the flags or range-flags file

```python
### ht 
df_sub = df.sample(1000, random_state=42).copy()
is_star = df_sub ['CH4_flag_a'] == '*'
is_f = df_sub['CH4_flag_a'] == 'F'
is_normal = ~(is_star | is_f)

plt.figure(figsize=(10,6))

# plot normal data points
plt.scatter(
    df_sub .loc[is_normal, 'year'],
    df_sub .loc[is_normal, 'CH4_area'],
    color = 'green',
    marker = 'o',
    label='Normal',
)
# plot '*' data points
plt.scatter(
    df_sub.loc[is_star, 'year'],
    df_sub.loc[is_star, 'CH4_area'],
    color = 'red',
    marker = '*',
    label = 'Flagged (*)'
)
# plot 'F' data points
plt.scatter(
    df_sub.loc[is_f, 'year'],
    df_sub.loc[is_f, 'CH4_area'],
    color = 'orange',
    marker = '^',
    label = 'Flagged (F)'
)

plt.xlabel('Year'),
plt.ylabel('CH4_area')
plt.title('CH4 Area Distribution by Year (Random Sampled, N = 1000)')
plt.legend()
plt.grid()
```

```python
df_sub = df.sample(1000, random_state=42).copy()
is_star = df_sub ['CH4_flag_ht'] == '*'
is_f = df_sub['CH4_flag_ht'] == 'F'
is_normal = ~(is_star | is_f)

plt.figure(figsize=(10,6))

# plot normal data points
plt.scatter(
    df_sub .loc[is_normal, 'year'],
    df_sub .loc[is_normal, 'CH4_ht'],
    color = 'green',
    marker = 'o',
    label='Normal',
)

# plot '*' data points
plt.scatter(
    df_sub.loc[is_star, 'year'],
    df_sub.loc[is_star, 'CH4_ht'],
    color = 'red',
    marker = '*',
    label = 'Flagged (*)'
)

# plot 'F' data points
plt.scatter(
    df_sub.loc[is_f, 'year'],
    df_sub.loc[is_f, 'CH4_ht'],
    color = 'orange',
    marker = '^',
    label = 'Flagged (F)'
)

plt.xlabel('Year'),
plt.ylabel('CH4_ht')
plt.title('CH4 Height Distribution by Year (Random Sampled, N = 1000)')
plt.legend()
plt.grid()
```

**Observation**
1. **`F`** : Mainly concentrated on extreme outliers as **zero values** or **siginificant outliers**
2. **`*`** : Mostly distributed within the **normal range of the main data**
