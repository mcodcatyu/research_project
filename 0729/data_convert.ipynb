```python
def split_flag (val):
    if pd.isna(val):
        return np.nan, np.nan
    
    val_str = str(val).strip()
    # check if last val is a character(Flag)
    if val_str[-1].isalpha() or val_str[-1] == '*':
        c_val = float(val_str[:-1])
        flag_val = val_str[-1]
    else:
        c_val = float(val_str) # normal value, then direcly convert to float
        flag_val = " " # no flag, give it a space
    
    return c_val, flag_val
```

```python
import pandas as pd
import numpy as np

file_path = "../data/raw/tac_ch4.txt"

df_tac = pd.read_csv(file_path, skiprows=1, sep=r'\s+', header=1) 
# skip first row, txt is space seperated

df_tac['C'], df_tac['ch4_flag'] = zip(*df_tac['C'].map(split_flag))


feature_name_tacch4 = [
  'date', 'time', 'type', 'sample', 'standard', 'port', 
 'dry', 'wet', 'stdev', 'std_rep', 'std_stdev',
  'target_error','Cdrift', 'C', 'N', 'Nfiltered',
  'cycle_time', 'h2o', 'h2o_stdev', 'cavity_press', 'cavity_press_stdev', 
  'cavity_temp', 'cavity_temp_stdev', 'das_temp', 'etalon_temp', 
  'warmbox_temp', 'outlet_valve','flag'
]
df_tac.columns = feature_name_tacch4
```

```python
import pandas as pd 
import polars as pl
import time
df_taco2= pd.read_csv("../data/processed/tac_co2_formatted_v1.csv")
flag_cov = {
    ' ': 0,
    'F':1,
    'x':0,
    '*': 1
}
df_taco2['label2'] = (df_taco2['co2_flag'].fillna(' ')).map(flag_cov).astype(int)
df_taco2['date'] = df_taco2['date'].astype(str).str.zfill(6)
df_taco2['time'] = df_taco2['time'].astype(str).str.zfill(6)


df_taco2['datetime_str'] = df_taco2['date'] + df_taco2['time']

df_taco2['datetime'] = pd.to_datetime(df_taco2['datetime_str'], format='%y%m%d%H%M%S')

df_taco2['datetime'] = pd.to_datetime(df_taco2['datetime'])

df_taco2['year'] =df_taco2['datetime'].dt.year

df_taco2.to_parquet('../data/processed/tac_co2_formatted_v2_label2.parquet', index=True)
```
