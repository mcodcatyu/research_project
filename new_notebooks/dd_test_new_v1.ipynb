```python
import pandas as pd
import numpy as np
import seaborn as sns
sns.set_theme(style='darkgrid')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
raw_data = pd.read_csv('../data/processed/mhd_ch4_simple_v1.csv')
raw_data.head()
```

```python
data = raw_data.copy()
data = data.drop(data[data['year']==2026].index)
data.tail()
```

```python
def types_convert(type_name):
    """
    This fnuction is used for type classification
    """
    t = str(type_name).lower().strip()

    if 'cal' in t:
        return 'cal'
    elif 'std' in t:
        return 'std'
    elif 'air' == t:
        return 'air'
    elif 'tank' in t:
        return 'tank'
    elif 'blank' in t:
        return 'blank'
    else:
        return 'other'
    
    
```

```python
data['type_norm'] = data['type'].apply(types_convert)
data['type_norm'].unique()
```

```python
data.tail()
```

```python
data = data.sort_values('datetime').reset_index(drop=True) # order by time
data.tail()
```

```python
data['datetime'] = pd.to_datetime(data['datetime'])
data['time_diff'] = data['datetime'].diff() # calculate the time interval with theprevious record
data.tail()
```

```python
data = data.sort_values(['type_norm', 'datetime']).reset_index(drop=True)

data['time_diff'] = data.groupby('type_norm')['datetime'].diff()
new_session = (data['time_diff'] > pd.Timedelta(days=30)) | (data)['time_diff'].isna()

data['session_id'] = data['type_norm'].astype(str) + '_ss_' + new_session.groupby(data['type_norm']).cumsum().astype(str)

data.tail()
```

```python
data['session_id'].unique()
```

```python
feature_cols = [
    #'sample'	,'standard',	
    #'port'	,'ht',
    'tmod'	,'tamb',
    'pflow', 'psamp',
    'pamb'	,
    'CH4_rt',	'CH4_w',	'CH4_ht'	,
    'CH4_area',	'CH4_skew'	,
    'CH4_start_time',	'CH4_end_time'	,'CH4_start_level',	'CH4_end_level',

    #'is_ht_999', 'is_area_999', 'is_rt_zero', 'is_w_zero',
    'duration', 
    'ht_area_ratio',
    #'type_air', 'type_blank', 'type_cal', 'type_other', 'type_std', 'type_tank',
    #'year', 'month', 'day', 'hour', 'minute', 'seconds'
]
```

```python
###handle NAN valuesmedian fill nan
for col in feature_cols:
    # first: fill in with previous value
    data[col] = data.groupby('session_id')[col].transform(lambda x: x.ffill())
    #second: median
    expanding_median = data.groupby('session_id')[col].transform(
        lambda x: x.expanding(min_periods=1).median()
    )
    data[col] = data[col].fillna(expanding_median)
    # third: whole type median
    type_median = data.groupby('type_norm')[col].transform('median')
    data[col] = data[col].fillna(type_median)
    # fourth: 0
    data[col] = data[col].fillna(0)
```

```python
for n, g in data.groupby('type_norm'):
    print(f'Type: {n}')
    print(f' -> Total data group:{len(g)}')
    print(f' -> Session number: {g["session_id"].nunique()}')
    print(f' -> earliest time: {g["datetime"].min()} ~ late time: {g['datetime'].max()}\n')

    broken_points = data[data['time_diff'] > pd.Timedelta(days=30)][['type_norm', 'datetime', 'time_diff']].head(15)
    print("broken time point")
    print(broken_points)
    print("------------------")
```

### residual calculation

```python
data = data.set_index('datetime')

for col in feature_cols:
    drift_values=data.groupby('session_id')[col].transform(
        lambda x: x.rolling(window='30D', min_periods=1).mean()
    ) # average

    data[f'{col}_drift_30_days'] = drift_values
    data[f'{col}_residual'] = data[col] - data[f'{col}_drift_30_days']
data = data.reset_index()
data.head()
```

```python
data['session_age'] = data.groupby('session_id')['datetime'].transform(
    lambda x: (x - x.min()).dt.total_seconds() / (24*60*60)
)
```

```python
model_features = [#'port', 'ht',
   #'sample'	,'standard',	
    #'port'	,'ht',
    'tmod'	,'tamb',
    'pflow', 'psamp',
    'pamb'	,
    'CH4_rt',	'CH4_w',	'CH4_ht'	,
    'CH4_area',	'CH4_skew'	,
    'CH4_start_time', 'CH4_end_time', 'CH4_start_level', 'CH4_end_level',

    #'is_ht_999', 'is_area_999', 'is_rt_zero', 'is_w_zero',
    'duration', 
    'ht_area_ratio',
    #'type_air', 'type_blank', 'type_cal', 'type_other', 'type_std', 'type_tank',
    #'year', 'month', 'day', 'hour', 'minute', 'seconds'

       #'type_norm', 'time_diff','session_id',
    'tmod_drift_30_days',
       'tmod_residual', 'tamb_drift_30_days', 'tamb_residual',

       'pflow_drift_30_days', 'pflow_residual', 'psamp_drift_30_days',
       'psamp_residual', 'pamb_drift_30_days', 'pamb_residual',

       'CH4_rt_drift_30_days', 'CH4_rt_residual', 'CH4_w_drift_30_days',
       'CH4_w_residual', 'CH4_ht_drift_30_days', 'CH4_ht_residual',

       'CH4_area_drift_30_days', 'CH4_area_residual', 'CH4_skew_drift_30_days',
       'CH4_skew_residual', 'CH4_start_time_drift_30_days',
       'CH4_start_time_residual', 'CH4_end_time_drift_30_days',
       'CH4_end_time_residual', 'CH4_start_level_drift_30_days',
       'CH4_start_level_residual', 'CH4_end_level_drift_30_days',
       
       'CH4_end_level_residual', 'duration_drift_30_days',
       'duration_residual', 'ht_area_ratio_drift_30_days', 'ht_area_ratio_residual', 'session_age']
```

```python
from sklearn.preprocessing import MinMaxScaler
import torch
from torch.utils.data import DataLoader, TensorDataset

X_raw = data[model_features].values

#===== scaler

scaler = MinMaxScaler() # for session_age 
X_scaled = scaler.fit_transform(X_raw)

X_all = torch.tensor(X_scaled, dtype=torch.float32)
y_all = data['label1'].values

normal_mask = (data['label1'] == 0).values
X_train = X_all[normal_mask]

print(f"Training shape:{X_train.shape}")
print(f"All data shape:{X_all.shape}")

train_dataset = TensorDataset(X_train)
train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True, num_workers=0)
```

```python
import torch.nn as nn
class GCAutoencoder(nn.Module):
    def __init__(self, input_dim=49):
        super(GCAutoencoder, self).__init__()
        # 49 -> 32 -> 16 -> 8
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16,8)
        )
        # decoder: 8-> 16 -> 32 -> 49
        self.decoder = nn.Sequential(
            nn.Linear(8,16),
            nn.ReLU(),
            nn.Linear(16,32),
            nn.ReLU(),
            nn.Linear(32, input_dim)
        )

    def forward(self, x):
        embedding = self.encoder(x)
        reconstructed = self.decoder(embedding)
        return reconstructed, embedding
        
```

```python
import torch.nn as nn
model = GCAutoencoder(input_dim=X_all.shape[1])
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)
```

```python
device
```

```python
import time

EPOCHS = 30
loss_history = []

total_samples = len(X_train)

print("Training ...")

model.train()

for epoch in range(EPOCHS):
    start_time = time.time()
    running_loss  = 0.0

    for batch in train_loader:
        inputs = batch[0].to(device)

        optimizer.zero_grad()

        reconstructed, embedding = model(inputs)

        loss = criterion(reconstructed, inputs)

        loss.backward()
        optimizer.step()

        running_loss +=loss.item() * inputs.size(0)
        
    epoch_loss = running_loss / len(X_train)

    loss_history.append(epoch_loss)

    elapsed = time.time() - start_time
    print(f"Epoch [{epoch+1:02d}/{EPOCHS}] | LOSS: {epoch_loss:.6f} | TIME COMSUME: {elapsed:.2f} s")
print("Finished")
```

```python
import matplotlib.pyplot as plt

plt.plot(loss_history, label='Train Loss')
plt.title('Autoencoder Training Progress')
plt.xlabel('Epochs')
plt.ylabel('MSE Loss')
plt.legend()
plt.show()
```

```python
model.eval()
with torch.no_grad():

    inputs_all = X_all.to(device)
    reconstructed_all, embedding_all = model(inputs_all)

    anomaly_scores = torch.mean((reconstructed_all -inputs_all) ** 2, dim=1).cpu().numpy()

results_data = data.sort_values('datetime').reset_index(drop=True).copy()
results_data['anomaly_score'] = anomaly_scores[data.sort_values('datetime').index]

print("Normal data ave MSE:", results_data[results_data['label1'] == 0]['anomaly_score'].mean() )
print("Abnormal data ave MSE:", results_data[results_data['label1'] == 1]['anomaly_score'].mean() )
```

```python
scores_for_threshold = results_data['anomaly_score'].copy()
scores_for_threshold[results_data['label1'] == 1] = np.nan

results_data['threshold_dynamic'] = (
    scores_for_threshold.rolling(window=600, min_periods=1, center=True).quantile(0.99)
)

results_data['threshold_dynamic'] = results_data['threshold_dynamic'].ffill().bfill()
```
