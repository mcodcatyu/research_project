```python
from statsmodels.tsa.seasonal import MSTL
def stl_and_plot(data, feature):
        mstl = MSTL(data[feature], periods=(36*7*4, 13140))
        result = mstl.fit()

        resid = result.resid
        # set threshold
        threshold = 3 * resid.std()
        #plotting

        fig, axes = plt.subplots(3, 1,figsize=(10,8))
        axes[0].plot(result.trend, color='red')
        axes[0].set_title('Drift')

        axes[1].plot(resid, color='green')
        axes[1].axhline(3*threshold, color='red')
        axes[1].axhline(-3*threshold, color='red')
        axes[1].set_title('Residual and threshold-Anomaly')

        axes[2].plot(data[feature])
        axes[2].set_title("original data")
        plt.tight_layout()
        plt.savefig("image/anomaly_plots_stl.png")
        plt.show()

```

```python
feature_cols = [
    #'sample'	,'standard',	
    #'port'	,'ht',
    'tmod'	,'tamb',
    'pflow', 'psamp',
    'pamb'	,
    'CH4_rt',	'CH4_w',	'CH4_ht'	,
    'CH4_area',	'CH4_skew'	,
    'CH4_start_time',	'CH4_end_time', 'CH4_start_level',	'CH4_end_level',

    #'is_ht_999', 'is_area_999', 'is_rt_zero', 'is_w_zero',
    'duration', 
    #'type_air', 'type_blank', 'type_cal', 'type_other', 'type_std', 'type_tank',
    #'year', 'month', 'day', 'hour', 'minute', 'seconds'
]
```

```python
one_year_data = data[(data['year']==2000)&(data['label1']==0)&(data['type']=='air')]
stl_and_plot(one_year_data,'CH4_rt')
```

```python
one_year_data = data[(data['year']==2000)&(data['label1']==0)&(data['type']=='std')]
stl_and_plot(one_year_data,'CH4_rt')
```
