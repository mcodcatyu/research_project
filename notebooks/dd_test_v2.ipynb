##### split air type and std to see if there is a difference in the residuals

```python
# one day 36 points
window_list = {
    '0.5_days': 18,
    '1_days': 36,
    '7_days': 36 * 7,
    '14_days': 36 * 14,
    '21_days': 36 * 21,
    '30_days': 36*30,
    '60_days': 36 * 60,
    '90_days': 36 * 90,
    '120_days': 36 * 120,
    '150_days': 36 * 150,
    '300_days': 36 * 300,
    '360_days': 36 * 360,
    '720_days':36*720
}
```

```python
def residual_cal(df, feature_cols, window_list):
    
    for f in feature_cols:
        results = {}
        print(f"Feature  {f}\n")
        for n, w in window_list.items():
            rolling_drift= df[f].rolling(window=w, min_periods=1, center=True).mean()

            residual = df[f] - rolling_drift

            #mad = np.median(np.abs(residual - np.median(residual)))
            rmse= np.sqrt(np.mean(residual ** 2))
            results[n] = {
                'rmse':rmse,
                'residual': residual
            }

            print(f"Window Size:{n:11s} | residual rmse score: {rmse:.10f}")

        best_window = min(results, key=lambda k: results[k]['rmse']) # find the minimal rmse
        print(f"\n best roling window size: {best_window}\n")
```

#### create residual columns, data preparing

```python
# the latest year data not used
data = raw_data.copy()
data = data.drop(data[data['year']==2026].index)
data.tail()
```

```python
# 
def types_convert(type_name):
    """
    This fnuction is used for type classification
    """
    t = str(type_name).lower().strip()

    if 'cal' in t:
        return 'cal'
    elif 'std' in t:
        return 'std'
    elif 'air' == t:
        return 'air'
    elif 'tank' in t:
        return 'tank'
    elif 'blank' in t:
        return 'blank'
    else:
        return 'other'
    
```

```python
data['type_norm'] = data['type'].apply(types_convert)
data['type_norm'].unique()
```

```python
data = data.sort_values('datetime').reset_index(drop=True) # order by time
data.head()
```

```python
data['datetime'] = pd.to_datetime(data['datetime'])
data['time_diff'] = data['datetime'].diff() # calculate the time interval with theprevious record
data.head()
```

```python
data = data.sort_values(['type_norm']).reset_index(drop=True)

data['time_diff'] = data.groupby('type_norm')['datetime'].diff()

data.head()
```

```python
data['session_id'] = (data['time_diff'] > pd.Timedelta(days=30)).cumsum() # 7 days as a window session
data['session_id'].unique()
```

```python
# calculate the mean during the window period, and keep within the same type
final_window = 36 * 7
for col in feature_cols :
    data[f'{col}_drift_7_days'] = data.groupby(['session_id', 'type_norm'])[col].transform(
        lambda x: x.rolling(window=final_window, min_periods=1, center=False).mean())
    data[f'{col}_residual'] = data[col] - data[f'{col}_drift_7_days']
```

```python
# type one-hot coding
data = pd.get_dummies(data, columns=['type_norm'], drop_first=False, dtype=float)
print(data.columns)
```

```python
import torch
df_normal = data[data['label1'] == 0].copy()

X_train = torch.tensor(df_normal[all_features].values, dtype=torch.float32)

X_all = torch.tensor(data[all_features].values, dtype=torch.float32)
y_all = data['label1'].values

print(f"Training shape:{X_train.shape}")
print(f"All data shape:{X_all.shape}")
```

```python
X_tensor = X_train.to(torch.float32)
```

```python
from torch.utils.data import TensorDataset, DataLoader

BATCH_SIZE = 2048

train_dataset = TensorDataset(X_tensor)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)

print(f"BATCH {len(train_loader)}")
```

```python
import torch.nn as nn
class GCAutoencoder(nn.Module):
    def __init__(self, input_dim=26):
        super(GCAutoencoder, self).__init__()

        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 16),
            nn.ReLU(),
            nn.Linear(16, 8),
            nn.ReLU(),
            nn.Linear(8,3)
        )

        self.decoder = nn.Sequential(
            nn.Linear(3,8),
            nn.ReLU(),
            nn.Linear(8,16),
            nn.ReLU(),
            nn.Linear(16, input_dim)
        )

    def forward(self, x):
        embedding = self.encoder(x)

        reconstructed = self.decoder(embedding)
        return reconstructed, embedding
        
model = GCAutoencoder(input_dim=38)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

```

```python
import torch.nn as nn

model = GCAutoencoder(input_dim=38)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)
```

```python
device
```

```python
import time

EPOCHS = 30
loss_history = []

total_samples = len(train_dataset)

print("Training ...")

model.train()

for epoch in range(EPOCHS):
    start_time = time.time()
    running_loss  = 0.0

    for batch in train_loader:
        inputs = batch[0].to(device)

        optimizer.zero_grad()

        reconstructed, embedding = model(inputs)

        loss = criterion(reconstructed, inputs)

        loss.backward()
        optimizer.step()

        running_loss +=loss.item() * inputs.size(0)
    epoch_loss = running_loss / len(X_train)
    loss_history = []
    loss_history.append(epoch_loss)

    elapsed = time.time() - start_time
    print(f"Epoch [{epoch+1:02d}/{EPOCHS}] | LOSS: {epoch_loss:.6f} | TIME COMSUME: {elapsed:.2f} s")
print("Finished")

```

```python
model.eval()
with torch.no_grad():

    X_all_device = X_all.to(device)
    reconstructed_all, embedding_all = model(X_all_device)

    mse_per_row = torch.mean((reconstructed_all - X_all_device) ** 2, dim=1).cpu().numpy()

data['anomaly_score'] = mse_per_row
print("Normal data ave MSE:", data[data['label1'] == 0]['anomaly_score'].mean() )
print("Abnormal data ave MSE:", data[data['label1'] == 1]['anomaly_score'].mean() )

```
