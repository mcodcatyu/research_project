```python
import pandas as pd
from sklearn import datasets
    
from evidently import Dataset
from evidently import DataDefinition
from evidently import Report
from evidently.presets import DataDriftPreset, DataSummaryPreset
```

```python
schema = DataDefinition(
    numerical_columns=[	'port',	'ht',	'pflow'	,
                       'psamp',	'pamb',	'CH4_rt',	'CH4_w'	,
                       'CH4_ht'	,'CH4_area'	,'CH4_skew'	,
                       'CH4_start_time'	,'CH4_end_time',	'CH4_start_level',	
                       'CH4_end_level'	,'duration',	'ht_area_ratio'],

    categorical_columns=['type_air'	,'type_blank'	,'type_cal'	,'type_other',	
                       'type_std', 'type_tank', 'port'],
    )
```

```python
reports = {}
for i in range(len(years)-1):
    y1 = years[i]
    y2 = years[i+1]

    reference =  Dataset.from_pandas(
        pd.DataFrame(df[df['year'] == y1].drop(columns=['tamb','tmod','year', 'month', 'day',	'hour', 'minute', 'seconds'])),
        data_definition=schema
    )
    current = Dataset.from_pandas(
        pd.DataFrame(df[df['year'] == y2].drop(columns=['tamb','tmod','year', 'month', 'day',	'hour', 'minute', 'seconds'])),
        data_definition=schema
    )

    report = Report([
    DataDriftPreset(
        num_method='wasserstein',
        num_threshold=0.1
    ) 
])

    my_eval = report.run(reference, current)
    my_eval.save_html(f'html/{y1}to{y2}.html')
```

```python
import pandas as pd
from evidently import Report
from evidently import Dataset, DataDefinition
from evidently.descriptors import Sentiment, TextLength, Contains
from evidently.presets import TextEvals, DataDriftPreset

drift_scores = []

for i in range(len(years)-1):
    y1 = years[i]
    y2 = years[i+1]

    reference = df[df['year'] == y1].drop(columns=['year'])
    current = df[df['year']==y2].drop(columns=['year'])

    report =Report([DataDriftPreset()])
    report.run(reference_data=reference, current_data=current)

    result_dict = report.dict()
    drift_detected = result_dict['metrics'][0]['result']['dataset_drift']
    drift_scores.append((y1, y2, drift_detected))
print(drift_scores)
```

```python
data_columns = ColumnMapping()
data_columns.numerical_features = ['weather', 'temp', 'atemp', 'humidity', 'windspeed']
```

```python
from evidently import Report
from evidently.presets import DataDriftPreset

drift_scores = []

for i in range(len(years)-1):
    y1 = years[i]
    y2 = years[i+1]

    reference = df[df['year'] == y1].drop(columns=['year'])
    current = df[df['year']==y2].drop(columns=['year'])

    data_drift_report = Report(metrics=[DataDriftPreset()])
    data_drift_report.run(reference_data=reference, current_data=current column_mapping=data_columns)
    report = data_drift_report.as_dict()
    report =Report([DataDriftPreset()])

    report.run(reference_data=reference, current_data=current)

    drift_result = report.metrics[0].get_result()
    drift_detected = drift_result.dataset_drift
    drift_scores.append((y1, y2, drift_detected))
print(drift_scores)
```
