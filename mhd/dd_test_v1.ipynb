```python
import pandas as pd
import numpy as np
import seaborn as sns
sns.set_theme(style='darkgrid')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
raw_data = pd.read_csv('../data/processed/mhd_ch4_simple_v1.csv')
raw_data.head()
```

```python
data = raw_data.copy()
data = data.drop(data[data['year']==2026].index)
data.tail()
```

```python
def types_convert(type_name):
    """
    This fnuction is used for type classification
    """
    t = str(type_name).lower().strip()

    if 'cal' in t:
        return 'cal'
    elif 'std' in t:
        return 'std'
    elif 'air' == t:
        return 'air'
    elif 'tank' in t:
        return 'tank'
    elif 'blank' in t:
        return 'blank'
    else:
        return 'other'
    
```

```python
data['type_norm'] = data['type'].apply(types_convert)
data['type_norm'].unique()
```

```python
data = data.sort_values('datetime').reset_index(drop=True) # order by time
data.head()
```

```python
data['datetime'] = pd.to_datetime(data['datetime'])
data['time_diff'] = data['datetime'].diff() # calculate the time interval with theprevious record
data.head()
```

```python
data = data.sort_values(['type_norm', 'datetime']).reset_index(drop=True)

data['time_diff'] = data.groupby('type_norm')['datetime'].diff()
new_session = (data['time_diff'] > pd.Timedelta(days=30)) | (data)['time_diff'].isna()

data['session_id'] = data['type_norm'].astype(str) + '_ss_' + new_session.groupby(data['type_norm']).cumsum().astype(str)

data.head()
```

```python
data['session_id'].unique()
```

```python
feature_cols = [
    #'sample'	,'standard',	
    #'port'	,'ht',
    'tmod'	,'tamb',
    'pflow', 'psamp',
    'pamb'	,
    'CH4_rt',	'CH4_w',	'CH4_ht'	,
    'CH4_area',	'CH4_skew'	,
    'CH4_start_time',	'CH4_end_time'	,'CH4_start_level',	'CH4_end_level',

    #'is_ht_999', 'is_area_999', 'is_rt_zero', 'is_w_zero',
    'duration', 
    'ht_area_ratio',
    #'type_air', 'type_blank', 'type_cal', 'type_other', 'type_std', 'type_tank',
    #'year', 'month', 'day', 'hour', 'minute', 'seconds'
]
```

```python
###handle NAN valuesmedian fill nan
for col in feature_cols:
    # first: fill in with previous value
    data[col] = data.groupby('session_id')[col].transform(lambda x: x.ffill())
    #second: median
    expanding_median = data.groupby('session_id')[col].transform(
        lambda x: x.expanding(min_periods=1).median()
    )
    data[col] = data[col].fillna(expanding_median)
    # third: whole type median
    type_median = data.groupby('type_norm')[col].transform('median')
    data[col] = data[col].fillna(type_median)
    # fourth: 0
    data[col] = data[col].fillna(0)
```

```python
data['datetime'] = pd.to_datetime(data['datetime'])
```

### residual calculation

```python
data = data.set_index('datetime')

for col in feature_cols:
    drift_values=data.groupby('session_id')[col].transform(
        lambda x: x.rolling(window='30D', min_periods=1).mean()
    ) # average

    data[f'{col}_drift_30_days'] = drift_values
    data[f'{col}_residual'] = data[col] - data[f'{col}_drift_30_days']
data = data.reset_index()
data.head()
```

```python
data['session_age'] = data.groupby('session_id')['datetime'].transform(
    lambda x: (x - x.min()).dt.total_seconds() / (24*60*60)
)
```

```python
import torch.nn as nn
class GCAutoencoder(nn.Module):
    def __init__(self, input_dim=26):
        super(GCAutoencoder, self).__init__()

        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 16),
            nn.ReLU(),
            nn.Linear(16, 8),
            nn.ReLU(),
            nn.Linear(8,3)
        )

        self.decoder = nn.Sequential(
            nn.Linear(3,8),
            nn.ReLU(),
            nn.Linear(8,16),
            nn.ReLU(),
            nn.Linear(16, input_dim)
        )

    def forward(self, x):
        embedding = self.encoder(x)

        reconstructed = self.decoder(embedding)
        return reconstructed, embedding
        
model = GCAutoencoder(input_dim=38)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```
