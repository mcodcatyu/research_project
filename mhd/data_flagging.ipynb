```python
# chec ht = 0 and c is nan
data = raw_data.copy()
data= data.drop(data[data['year']==2026].index)
data['is_ht_zero_and_C_Nan'] = ((data['CH4_ht'] == 0)&(data['CH4_C'].isna())).astype(int)
data.head()
```

```python
data['previous_point_type'] = data['type'].shift(1)
data['next_point_type'] =data['type'].shift(-1)
```

```python
# prvious is std, next is air
first_index = data[((data['is_ht_zero_and_C_Nan']==1)&((data['type'] == 'std')|(data['type'] == 'air')) & (data['previous_point_type']=='std')&((data['next_point_type']=='air')))].index.tolist()
first_index
```

```python
# previous is air,next is std
second_index = data[((data['is_ht_zero_and_C_Nan']==1)&((data['type'] == 'std')|(data['type'] == 'air'))&(data['previous_point_type']=='std')&((data['next_point_type']=='air')))].index.tolist()
second_index
```

```python
data[((data['is_ht_zero_and_C_Nan']==1)&(data['previous_point_type']=='std')&((data['type'] == 'std')|(data['type'] == 'air'))&((data['next_point_type']=='air')))]
```

```python
import numpy as np
```

```python
data['is_bad_std'] = ((data['type'] == 'std') & (data['is_ht_zero_and_C_Nan']==1)).astype(int)

data['is_normal_std'] =((data['type'] == 'std') & (data['is_ht_zero_and_C_Nan']==0)).astype(int)

data['error_state'] = (np.where(data['is_bad_std'], True,
                                np.where(data['is_normal_std'], False, np.nan))).astype(float)

# detect error serires
data['error_state'] = data['error_state'].ffill()

data['is_protential_flagged_air'] = ((data['error_state'] == True) & (data['type'] == 'air')).astype(int)

#print(data[data['is_protential_flagged_air'] == 1])
(data['is_protential_flagged_air'] == 1).sum()
#potential error air data
print(((data['is_protential_flagged_air'] == 1)&(data['label1'] == 0)).sum())

data[(data['is_protential_flagged_air'] == 1)&(data['label1'] == 0)&(data['is_ht_zero_and_C_Nan']==1)]
```
